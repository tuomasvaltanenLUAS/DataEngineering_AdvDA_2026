### Dataset


```python
import seaborn as sns
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler, OneHotEncoder
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score, adjusted_rand_score

import warnings
warnings.filterwarnings("ignore")

# Loading a synthetic dataset
df = pd.read_csv("clusterable_dataset.csv")
df
```

### Using agglomerative clustering





```python
# X, y split
X = df.drop(["segment", "merchant_id"], axis=1)
y = df["segment"]

# Fitting, modeling, and adding a 'cluster' column.
model = AgglomerativeClustering(n_clusters=3, linkage="ward")
clusters = model.fit_predict(X)
df["cluster"] = clusters

# Setting up metrics -- Silhouette and ARI
silhouette = silhouette_score(X,clusters)
ari_score = adjusted_rand_score(y, clusters)

# Sillhouette score works for both REGRESSION and CLASSIFICATION
# Range : 0-1 ,  The closer to 1 the better 
print(f"Silhouette Score: {silhouette:.4f}")

# ARI score is only useful for classification -- NOT REGRESSING
# Range : 0-1 ,  The closer to 1 the better
print(f"Adjusted Rand Index (ARI): {ari_score:.4f}")
print("")
print("")

df
```
