### Cluster profiling - using assigned cluster values to create new variables

Cluster profiling might seem like it's target leakage, but we have to think about this carefully:
- We don't know the target value in real-time (with new data)
- ... but we do know the cluster value based on KMeans
- ... based on KMeans-based cluster value, we can also know the average target value for each cluster
- ... => no actual target leakage

See version 2 of this notebook for additional thoughts.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import seaborn as sns

# load data
df = pd.read_csv("../winequality-red.csv")
df
```

```python
# X/y -split
X = df.drop("quality", axis=1)

# most clustering algorithms require scaled values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# create clustering model and train it with our data
kmeans = KMeans(n_clusters=6, random_state=42)

# place the cluster value back to DataFrame
df['cluster'] = kmeans.fit_predict(X_scaled)

df.head(5)
```

```python
# let's see how the cluster values are spread out
df['cluster'].value_counts()
```

```python
# let's see how the clusters are spread around the qualities (target variable)

# for clustering attempts, it's recommended to lock down the random seed

# it's a fairly usable clustering, although not optima
# ideally the clustering should clearly separate different qualities in the
# target variable...
# currently, clusters 0 and 2 seem to emphasize quality 5 somewhat
# and also clusters 1, 4 and 5 emphaisize quality 6 
# then again, there's still overlap in all of these clusters
cluster_counts = df.groupby(['quality', 'cluster']).size().unstack(fill_value=0)
print(cluster_counts)
```

### Perform cluster profiling => calculate averages based on the created clusters

```python
# calculate the average quality for each cluster
cluster_quality_average = df.groupby('cluster')['quality'].mean()

# place the averages based on clusters back into the DataFrame
df['cluster_average_quality'] = df['cluster'].map(cluster_quality_average)

df
```

```python
# study the average quality for each cluster
# these averages are probably not that helpful for an ML model (mostly same averages)
# this implies => overlap and noise in data
df.groupby("cluster", as_index=False)['cluster_average_quality'].mean()
```
