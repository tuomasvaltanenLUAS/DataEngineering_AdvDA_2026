### Simple interaction feature - example - Manually with pandas

```python
import pandas as pd
import numpy as np

# it seems the data doesn't follow the UTF-8 format
# let's force the encoding to follow Latin1 (or ISO-8859-1)
df = pd.read_csv("../OnlineRetail.csv", encoding="ISO-8859-1")

# a very simple interaction feature by MULTIPLICATION
# you can use same logic with additions, substraction, division etc.
# just be careful you don't divide by zero

# the equations can be also more advanced, e.g.
# df['Sales'] = df['YearlySales'] / 12 * df['AddedTax'] depending on situation
df['Sales'] = df['Quantity'] * df['UnitPrice']
df
```
