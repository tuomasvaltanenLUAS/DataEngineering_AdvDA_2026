```python
# import torch
# pip install tabfairgan
from tabfairgan import TFG
```

```python
import pandas as pd
df = pd.read_csv("adult.csv")
```

```python
fairness_config = {
    'fair_epochs': 50,
    'lamda': 0.5,
    'S': 'sex',
    'Y': 'income',
    'S_under': ' Female',
    'Y_desire': ' >50K'
}

tfg = TFG(df, epochs=100, batch_size=256, device="cpu", fairness_config=fairness_config)
```

```python
tfg.train()
```

```python
fake_df = tfg.generate_fake_df(2000)
```
