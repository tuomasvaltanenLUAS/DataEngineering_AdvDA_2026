### Phik-matrix -> Associations

Associations measure the strength of assocation between two variables. While correlations only measure LINEAR connection, associations can also handle non-linear connections. See "associations_vs_correlations.ipynb" for a simple comparison where associations are usually more useful.

**NOTE!** You can see both the correlations and associations at the asme time if you use SweetViz or Dython! You can also have correlations/phik-matrix in ydata at the same time, if you configure it correctly.

```python
# if phik-matrix not found in pandas, try:
# pip install phik
import phik
import pandas as pd

# load data
df = pd.read_csv("../winequality-red.csv")

# compute phik-matrix values
phik_matrix = df.phik_matrix()

phik_matrix
```

```python
phik_matrix['quality'].sort_values(ascending=False)
```

Correlations for comparison

```python
# you can compare to correlations too
df.corr(numeric_only=True)
```

```python
df.corr(numeric_only=True)['quality'].sort_values(ascending=False)
```
