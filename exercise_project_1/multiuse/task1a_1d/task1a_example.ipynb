### Task 1a, inspect the datatypes for each column

Most common cases are date-columns being string/object, or all columns being in string -format (including numeric columns).

```python
# imports
import pandas as pd

# open dataset file, fix the file path if you use a different structure 
# in your exercises
df = pd.read_csv("../winequality-red.csv")

# see the datatypes
df.dtypes
```

```python
df.head()
```

### We can see that "date" is an "object", we should convert this to DateTime

Completely depends on your dataset, if anything needs to be changed.

```python
# just got this from Copilot
df["date"] = pd.to_datetime(df["date"],format="%Y%m%dT%H%M%S")
```

```python
# seems to work!
df['date'].dt.year
```
