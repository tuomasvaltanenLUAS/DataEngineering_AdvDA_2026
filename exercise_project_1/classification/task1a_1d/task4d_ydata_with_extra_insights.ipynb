### y-data with extra features

This versions adds phik-assocations to correlation matrix.
Also adds skewness alerts to the recomendations.

```python
from ydata_profiling import ProfileReport
from ydata_profiling.config import Settings
import pandas as pd
import numpy as np


df = pd.read_csv("../lifestyle_sustainability_data.csv")

# Create configuration
config = Settings()

# Set the skewness threshold to whichever you prefer.
config.vars.num.skewness_threshold = 2

# Enable correlations
config.correlations["pearson"].calculate = True
config.correlations["phi_k"].calculate = True

# Generate the profile report
report = ProfileReport(
    df,
    title='Sustainable Lifestyle Rating data, Data Engineering / Advanced Data Analytics, 2026',
    config=config
)

# Save report
report.to_file("ydata_report_classification_extra_features.html")
```
