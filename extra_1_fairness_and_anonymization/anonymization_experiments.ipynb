```python
import pandas as pd
import random
import string
import numpy as np

# load the data
original_df = pd.read_csv("Customers_data.csv")

# create an imaginary Salary so we also have some numeric column
original_df['Salary'] = np.random.randint(25000, 175000, size=len(original_df))

# since some of the anonymization methods are quite heavy and slow
# we'll try these with a smaller sample
original_df = original_df[0:1000]
original_df
```

**Experiment 1: Randomization**

```python
# define function that operates on entire dataframe
def randomize_values(df):
    for column in df.columns:
        if df[column].dtype == 'O': # check if column has object dtype
            df[column] = [''.join(random.choices(string.ascii_letters + string.digits, k=10)) for _ in range(len(df))] # generate a list of random strings
    return df

# apply function to dataframe
df_rand = randomize_values(original_df.copy())

# all values but numeric are now randomized
df_rand
```

**Experiment 2: Aggregation (summary, simplify, group by data)**

```python
# Anonymize the ages by grouping them into age ranges
bins = [0, 30_000, 50_000, 80_000, 120_000, 200_000]
labels = ['0-30K', '31-50K', '51-80K', '81-120K', '120K+']
df_rand['Salary'] = pd.cut(df_rand['Salary'], bins=bins, labels=labels)
```

**Experiment 4: Perturbation**

```python
def add_noise(df, column, std = None):
    if std == None:
        std = df[column].std()
    
    withNoise = df[column].add(np.random.normal(0, std, df.shape[0]))
    copy = df.copy()
    copy[column] = withNoise
    return copy


# adds a small noise to Salary so it's not exactly matching with real data
perturbation = add_noise(original_df, 'Salary', std=100)
perturbation
```
